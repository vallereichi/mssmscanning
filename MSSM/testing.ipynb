```python
import h5py
import numpy as np
```

```python
GAMBIT_v1 = "../data/version-9/FS_MSSM7atQ_diver_gm2_gambit.hdf5"
GAMBIT_v2 = "../data/version-10/FS_MSSM7atQ_diver_gm2_gambit.hdf5"
test_parameter = "#MSSM_spectrum @SpecBit::get_MSSM_spectrum_as_map::~chi0_2 Pole_Mass"

scan_v1 = h5py.File(GAMBIT_v1, "r")["MSSM"]
scan_v2 = h5py.File(GAMBIT_v2, "r")["MSSM"]

data_v1 = scan_v1[test_parameter][scan_v1[test_parameter + "_isvalid"][:] == 1]
data_v2 = scan_v2[test_parameter][scan_v2[test_parameter + "_isvalid"][:] == 1]
```
